# Import libs

```python
from driller.data.changepoints_generators.changepoint_generators import LinearSteps, SyntheticSinusoid
import numpy as np
import pandas as pd

import random
from datetime import datetime
```

# Define Data Class Generator

```python
class FileGenerator:
    def __init__(self, local_path: str = "/home/gishb/datasets/change_points/syth"):
        self.local_path = local_path

    def create_dataframe(self, data: np.ndarray) -> pd.DataFrame:
        return pd.DataFrame(data=data.T[:, [0, 1]], columns=["cps", "values"])

    def save_dataframe(self, data: np.ndarray, prefix: str, length_data: int, index: int):
        df = self.create_dataframe(data)
        timestamp = datetime.now()
        path = f"{self.local_path}/{length_data}_{timestamp.year}_{timestamp.month}_{timestamp.day}_{timestamp.hour}:{timestamp.minute}_{prefix}_data_{index}.csv"
        df.to_csv(path, index=False)
        
```

# Define global args

```python
N_SAMPLES = 1_000_00
LENGTH_INFO = 450
local_path = "/home/gishb/datasets/change_points/validation"
```

# Define Pipeline to generate data

```python
generator = FileGenerator(local_path=local_path)
```

```python
for i in range(N_SAMPLES):
    rand_val = random.randint(a=0, b=1)
    rand_cps = random.randint(a=1, b=10)
    if rand_val:
        data = LinearSteps(cps_number=rand_cps, length_data=LENGTH_INFO).get()
        generator.save_dataframe(data, prefix="LinearSteps", length_data=LENGTH_INFO, index=i)
    else: 
        data = SyntheticSinusoid(cps_number=rand_cps, length_data=LENGTH_INFO).get()
        generator.save_dataframe(data, prefix="SyntheticSinusoid", length_data=LENGTH_INFO, index=i)
```
